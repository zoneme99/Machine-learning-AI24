```python
from collections import Counter
import matplotlib.pyplot as plt

lista = list()
movies['genres'].apply(lambda x: lista.append(x.split('|')))
lista = [inner for outer in lista for inner in outer]
genres = dict(Counter(lista))
plt.xticks(rotation=90)
plt.bar(genres.keys(), genres.values())
```

```python
print(len(movies['genres'].unique()))
print(genres['Film-Noir'])
print(genres['IMAX'])
print(len(movies[movies['genres'] == 'Film-Noir']))
print(len(movies[movies['genres'] == 'IMAX']))
```
