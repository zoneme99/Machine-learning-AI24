```python
import matplotlib.pyplot as plt
chol = df['cholesterol']
normal = chol[chol == 1]
above = chol[chol == 2]
well_above = chol[chol == 3]
plt.pie([len(normal),len(above), len(well_above)], labels=['Normal', 'Above', 'Way Above'], colors=['green', 'orange', 'red'])
plt.title('Cholesterol')
plt.show()
```
